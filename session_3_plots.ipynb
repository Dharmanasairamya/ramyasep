**pie chart**
- autopact=%a.bf%%
-  first % is the formatting
-  a: how much minimum width
-  b: after point how many decimals
-  %% : % symbol
-  %1.1f%% : 40.2%
-  %2.2f%% : 423.23

```python
keys=visa_data['continent'].value_counts().keys()
values=visa_data['continent'].value_counts().values
```

```python
keys
```

```python
values
```

```python
#plt.pie(x,labels)
colors=['fuchsia','blue','green','yellow','purple','orange']

plt.pie(values,
        explode=[0.1 for i in range (6)],
        labels=keys,
        colors=colors,
        autopct='%1.2f%%',
        radius=0.8,
        startangle=40)
plt.savefig('continent_pie.jpg')
plt.show()
```

```python
#case-3: subplot
col=['continent','case_status']
keys1=visa_data['continent'].value_counts().keys()
value1=visa_data['continent'].value_counts().values
plt.figure(figsize=(18,3))
plt.subplot(1,2,1)
plt.bar(keys1,value1)


keys2=visa_data['case_status'].value_counts().keys()
value2=visa_data['case_status'].value_counts().values
plt.subplot(1,2,2)
plt.bar(keys2,value2)
```

```python
#case-3: using for loop
col=['continent','case_status']
plt.figure(figsize=(18,3))
for i in range(len(col)):
    keys=visa_data[col[i]].value_counts().keys()
    values=visa_data[col[i]].value_counts().values
    plt.subplot(1,2,i+1).bar(keys,values)
```
