```python
# #step-9:Step-9: Encoding methods
# map
# np.where
# LabelEncoder
# onehot encoder
```

```python
#map
telecom_data["Churn"] = telecom_data["Churn"].map({
    "Yes": 1,
    "No": 0
})
telecom_data
```

```python
#np.where
where=telecom_data["churn"] = np.where(
    telecom_data["churn"] == "Yes",
    1,
    0
)

where
```

```python
# label_encoder
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

telecom_data["gender"] = le.fit_transform(telecom_data["gender"])
```

```python
#step-10 :Scaling methods
# Z
# minmax scalar
```

```python
#z
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

telecom_data[num] = scaler.fit_transform(telecom_data[num])
```

```python
# min max
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

telecom_data[num] = scaler.fit_transform(telecom_data[num])
```
