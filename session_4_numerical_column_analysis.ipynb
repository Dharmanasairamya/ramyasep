**emperical rule**

- 68% of data available between u-1*sd to u+1*sd
- step-1: calculate mean of wage
- step-2: calculate sd of wage:sd
- step-3: 1b=mean-1*sd
- step-4: ub=mean+1*sd
- step-5: condi1=wage_data>1b
- step-6:condi2=wage_data<ub
- step-7:con=condi1 & condi2
- step-8: np.sum(con)

```python
wage_data=visa_data['prevailing_wage']
mean=wage_data.mean()
sd=wage_data.std()
ib=mean-1*sd
ub=mean+1*sd
condi1=wage_data>ib
condi2=wage_data<ub
condition=condi1 & condi2
print(np.sum(condition))
print(68*25480/100)
np.sum(condition)==68*(254800)/100
```

**capping**
- we can fill the outliers which is less than 25 p with 25p value
- we can fill the outliers which is less than 75 p with 75 p value

```python
wage_data=visa_data['prevailing_wage']
q1=np.percentile(wage_data,25)
q2=np.percentile(wage_data,50)
q3=np.percentile(wage_data,75)
iqr=q3-q1
lb=q1-1.5*iqr
ub=q3+1.5*iqr
visa_data['wage_data_cap']=wage_data.clip(lb,ub)
```

```python
wage_data=visa_data['prevailing_wage']
wage_data_q2=visa_data['wages_q2']
wage_data_cap=visa_data['wage_data_cap']
plt.figure(figsize=(15,4))
plt.subplot(1,3,1).boxplot(wage_data,vert=False)
plt.subplot(1,3,2).boxplot(wage_data_q2,vert=False)
plt.subplot(1,3,3).boxplot(wage_data_cap,vert=False)
plt.show()
```
